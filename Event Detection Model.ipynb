```python
from ultralytics import YOLO
model = YOLO("yolo11s.pt")

results = model.train(
    data=f"/content/Ambient-Intelligence-Classroom-1/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,

    patience=10,

    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,

    degrees=10,
    translate=0.1,
    scale=0.5,
    shear=2.0,
    fliplr=0.5,

    mosaic=1.0,
    mixup=0.1,

    workers=2,

    project="runs/detect",
    name="classroom_yolo11s",
    exist_ok=True,
    save=True
)
```

```python
from ultralytics import YOLO

# Load the best model
model = YOLO("/content/runs/detect/runs/detect/classroom_yolo11s/weights/best.pt")

# Validate the model
metrics = model.val(
    data=f"/content/Ambient-Intelligence-Classroom-1/data.yaml"
)

# Print metrics
print("================================")
print("Model Evaluation")
print("================================")

print(f"mAP50     : {metrics.box.map50:.4f}")
print(f"mAP50-95  : {metrics.box.map:.4f}")
print(f"Precision : {metrics.box.mp:.4f}")
print(f"Recall    : {metrics.box.mr:.4f}")
```
